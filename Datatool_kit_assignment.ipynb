14. Explain the difference between apply() and map() in Pandas?**

--> In Pandas, both `apply()` and `map()` are used to apply a function to a Series or DataFrame, but they differ in their usage and flexibility:

*   **`map()`:** This method is primarily used on a **Series** to substitute each value in the Series with another value. It can take a dictionary, a Series, or a function as an argument. When a dictionary or Series is used, it performs element-wise mapping based on the keys. When a function is used, it applies the function to each element of the Series. `map()` is designed for element-wise transformations.

*   **`apply()`:** This method can be used on both **Series** and **DataFrames**. It is more flexible than `map()` as it can apply a function along an axis of the DataFrame (row-wise or column-wise) or to a Series. When used on a DataFrame, `apply()` can operate on entire rows or columns as Series objects, allowing for more complex operations that involve multiple elements.

In essence, `map()` is specialized for element-wise transformations on a Series, while `apply()` is more general and can operate on Series or DataFrames along an axis, allowing for more complex operations.

18. Why is NumPy’s array slicing faster than Python’s list slicing?

-- > NumPy's array slicing is significantly faster than Python's list slicing due to fundamental differences in how these data structures are stored and handled in memory:
Contiguous Memory Allocation: NumPy arrays store elements of the same data type in contiguous blocks of memory. This allows for efficient access and manipulation by the CPU, as data can be loaded into cache and processed quickly. Python lists, on the other hand, store references (pointers) to objects that can be scattered across different memory locations, leading to less efficient memory access.
Homogeneous Data Types: NumPy arrays are designed for homogeneous data, meaning all elements must be of the same data type. This fixed data type allows NumPy to optimize memory usage and perform operations at compiled C speed, without the overhead of checking individual element types during operations. Python lists can store heterogeneous data types, requiring more overhead for type checking and object management.
Vectorization and Optimized C Code: NumPy leverages vectorization, which enables operations on entire arrays simultaneously rather than processing individual elements sequentially. These vectorized operations are implemented in highly optimized C code, leading to substantial performance gains compared to Python's interpreted loops and list operations.
Reduced Overhead per Element: Python lists store additional information with each element, such as type and reference count, adding overhead. NumPy arrays, due to their homogeneous nature and contiguous storage, have significantly less overhead per element, making them more memory-efficient and faster for numerical computations, especially with large datasets.

```python

#Create a line plot with multiple lines using Matplotlib

import matplotlib.pyplot as plt
import numpy as np

# Generate sample data for multiple lines
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.sin(x) + np.cos(x)

# Create the line plot
plt.figure(figsize=(10, 6)) # Optional: adjust figure size
plt.plot(x, y1, label='sin(x)')
plt.plot(x, y2, label='cos(x)')
plt.plot(x, y3, label='sin(x) + cos(x)')

# Add labels, title, and legend
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Line Plot with Multiple Lines using Matplotlib')
plt.legend() # Show the legend to identify the lines
plt.grid(True) # Optional: add a grid

# Display the plot
plt.show()
```

```python
# Generate a Pandas DataFrame and filter rows where a column value is greater than a threshold

import pandas as pd

# Create a sample DataFrame
data = {'col1': [10, 20, 30, 40, 50],
        'col2': [100, 150, 250, 350, 450]}
df = pd.DataFrame(data)

# Define the threshold
threshold = 30

# Filter rows where 'col1' is greater than the threshold
filtered_df = df[df['col1'] > threshold]

print("Original DataFrame:")
print(df)
print(f"\nDataFrame filtered where 'col1' > {threshold}:")
print(filtered_df)
```
